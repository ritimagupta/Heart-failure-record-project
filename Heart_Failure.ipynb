```python
import pandas as pd

df = pd.read_csv('/content/heart+failure+clinical+records.zip')

print(df.head())
print(df.tail())

print(df.shape)

print(df.info())

print(df.describe())



```

```python
#age distribution
age_distribution = df["age"].value_counts().sort_index()
print(age_distribution)

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df['age'],
         bins=10,
         color='lightgreen',
         edgecolor='black')

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
# Count records by death_event
df["DEATH_EVENT"].value_counts().plot(
    kind="bar",
    color=["skyblue","salmon"]
)

plt.title("Death Event Distribution")
plt.xlabel("DEATH_EVENT")
plt.ylabel("Count")
plt.show()
```

```python
import numpy as np

#Find mean, median, minimum and maximum for age, ejection_fraction and serum_creatinine

age_np = np.array(df["age"])
ef_np = np.array(df["ejection_fraction"])
sc_np = np.array(df["serum_creatinine"])

print(np.mean(age_np))
print(np.median(age_np))
print(np.min(age_np))
print(np.max(age_np))
```

```python
#ompare average ejection_fraction and serum_creatinine across death_event groups.
df.groupby("DEATH_EVENT")[
    ["ejection_fraction","serum_creatinine"]
].mean()
```

```python
#Count records with and without diabetes, high blood pressure and anaemia.

# Diabetes Count
print("Diabetes Counts:")
print(df["diabetes"].value_counts())

# High Blood Pressure Count
print("\nHigh Blood Pressure Counts:")
print(df["high_blood_pressure"].value_counts())

# Anaemia Count
print("\nAnaemia Counts:")
print(df["anaemia"].value_counts())
```
